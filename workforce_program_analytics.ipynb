# Workforce Program Analytics (Mock Data)

This notebook includes mock workforce program data, ETL, statistical analysis, and visualizations.

```

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
import random
```

```

# Settings
np.random.seed(42)
num_rows = 500
programs = ['Job Readiness', 'Technical Skills', 'Career Coaching']

# Generate mock data
data = pd.DataFrame({
    'Participant_ID': range(1, num_rows + 1),
    'Program': np.random.choice(programs, size=num_rows),
    'Start_Date': [datetime(2024,1,1) + timedelta(days=random.randint(0, 365)) for _ in range(num_rows)],
})

# Random durations and metrics
data['Duration_Days'] = np.random.randint(10, 60, size=num_rows)
data['End_Date'] = data['Start_Date'] + pd.to_timedelta(data['Duration_Days'], unit='d')
data['Hours_Trained'] = np.random.randint(10, 100, size=num_rows)
data['Completion_Status'] = np.random.choice(['Complete', 'Incomplete'], size=num_rows, p=[0.75, 0.25])
data['Outcome_Score'] = np.random.randint(50, 100, size=num_rows)

# Save mock dataset
data.to_csv('workforce_data_mock.csv', index=False)
data.head()
```

```

# Load data
data = pd.read_csv('workforce_data_mock.csv')

# Clean data
data['Start_Date'] = pd.to_datetime(data['Start_Date'])
data['End_Date'] = pd.to_datetime(data['End_Date'])
data['Program'] = data['Program'].str.title()
data['Hours_Trained'] = data['Hours_Trained'].fillna(data['Hours_Trained'].median())
data['Success'] = np.where(data['Outcome_Score'] > 80, 1, 0)
data.head()
```

```

# Summary statistics
summary = data.groupby('Program').agg({
    'Participant_ID': 'count',
    'Hours_Trained': 'mean',
    'Duration_Days': 'mean',
    'Success': 'mean'
}).reset_index()
summary.rename(columns={'Participant_ID': 'Participants', 'Success': 'Success_Rate'}, inplace=True)
summary
```

```

# Bar chart: Success rate per program
plt.figure(figsize=(8,5))
sns.barplot(x='Program', y='Success_Rate', data=summary)
plt.title('Program Success Rates')
plt.ylabel('Success Rate')
plt.ylim(0,1)
plt.show()

# Scatter plot: Hours trained vs success
plt.figure(figsize=(8,5))
sns.scatterplot(x='Hours_Trained', y='Success', data=data)
plt.title('Hours Trained vs Success')
plt.ylabel('Success (1=Yes, 0=No)')
plt.show()
```
